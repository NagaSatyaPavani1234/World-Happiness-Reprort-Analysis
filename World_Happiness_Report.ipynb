# **Data preprocessing**

```python
#counting the number of null values in each column of the dataframe
df.isnull().sum()
```

```python
#since there is no null values directly checking for duplicate values in the dataframe
df.duplicated().sum()
```

Adding a categorical column to the dataframe

```python
print("maximum happinesss is",df["Happiness Score"].max())
print("minimum happinesss is",df["Happiness Score"].min())
```

```python
#adding new  happiness_level categorical column to the dataframe
l=[]
for i in df["Happiness Score"]:
  if i<3:
    l.append("poor")
  elif i<=5:
    l.append("low")
  elif i<=6:
    l.append("average")
  else:
    l.append("good")
df["Happiness level"]=l
```

```python
df["Happiness level"].value_counts()
```

# **Data Analysis**

```python
#Which country has the highest happines score?
df.loc[df["Happiness Score"].idxmax()]["Country"]
```

```python
#Find the mean of Happiness score based on Happiness level
df.groupby(by="Happiness level")["Happiness Score"].mean().sort_values().reset_index()
```

```python
#How many countries fall in each happiness categiry
df["Happiness level"].value_counts().reset_index()
```

```python
#What are the top 10 happiest Countries
df.sort_values(by="Happiness Score").head(10)["Country"]
```

```python
#which countries has Happiness score higher than the average Happiness Score and display the rank also
AvgScore=df["Happiness Score"].mean()
df1=df[df["Happiness Score"]>AvgScore]
df1[["Country","Happiness Rank"]]
```

```python
#Which countries has high happiness and below average GDP
AvgGDP=df["Economy (GDP per Capita)"].mean()
cond1=df['Economy (GDP per Capita)']<AvgGDP
cond2=(df["Happiness level"]=='good')
df[(cond1 & cond2)]["Country"]
```

```python
#Which countries have above average life expectancy but below average happiness
AvgScore=df['Happiness Score'].mean()
AvgLife=df['Health (Life Expectancy)'].mean()
cond1=(df["Happiness Score"]<AvgScore)
cond2=(df['Health (Life Expectancy)']>AvgLife)
df[(cond1 & cond2)]["Country"]
```

```python
#how many countries has happiness scores above the median
medianScore=df['Happiness Score'].median()
print("Total number of have happiness scores above the median = ",df[df["Happiness Score"] > medianScore].shape[0])
```

```python
#Compare average GDP per capita for each happiness category
df.groupby("Happiness level")['Economy (GDP per Capita)'].mean()
```

```python
#Which factor is most correlated with happiness score?
df.corr(numeric_only=True)["Happiness Score"].sort_values(ascending=False)
```

# **Data Visualizations**






# Box PLot

```python

#Distribution of happiness Score
sns.boxplot(x=df["Happiness Score"])
plt.title("Distribution of Happiness Score")
plt.show()
```

# Observation

In Box plot  the median line is at the point 5.3, it means half of the countries happiness left below 5.3 and half of the countries happiness is above 5.3
Here the happiness score is distributed accross the range from 2.3 to 7.54
Here most of the scores lies between 4.5 to 6.3, so most of the countries has moderate happiness


# Correlation Heatmap

```python

# It is used to show the relationship (correlation) between numerical variables in a dataset.
plt.figure(figsize=(8,6))
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap="coolwarm")#annot=True results the values of correlation inside the cells
plt.title("Correlation Matrix")
plt.show()
```

# Observation

The heatmap indicates that GDP per capita, family support, and life expectancy are the strongest factors associated with happiness score, while generosity and corruption show weaker relationships.


# Bar Graph

```python

#Average Happiness Score by Happiness Level
sns.barplot(x="Happiness level", y="Happiness Score", data=df,width=0.4)
plt.title("Average Happiness Score by Happiness Level")
plt.xlabel("Happiness Level")
plt.ylabel("Happiness Score")
plt.show()
```

# Observation

From the above graph we can observe that happiness level is directly proportional to happiness score as score increases ,level also increases
The countries  having good happiness level has greater Happiness score
The countries having poor happiness level has lesser Happiness Score

# Line Plot

```python

# Plot the Happiness Score by Country Rank
sns.lineplot(x=df['Happiness Score'],y=df["Happiness Rank"],hue=df['Happiness level'])
plt.title('Happiness Score Trend Across Countries')
plt.show()
```

# Observation

As the happiness score increases, the happiness rank improves.
Countries in the good category have the highest scores, while those in the poor category have the lowest scores.

# Scatter Plot

```python

#Healthy Life Expectancy vs Happiness Score
sns.scatterplot(x=df['Health (Life Expectancy)'],y=df['Happiness Score'],hue=df['Happiness level'])
plt.title('Life ecpectancy vs Happiness Score')
plt.show()
```

# Obseravtion

The plot shows a clear trend where happiness scores increase as life expectancy improves.
Countries with better health conditions seem to report higher levels of happiness, while those with lower life expectancy tend to have lower happiness scores.

# Pie Chart

```python

#pie chart of Happiness levels
counts = df["Happiness level"].value_counts()
plt.pie(counts,
        labels=counts.index,
        autopct="%1.1f%%",
        colors=["red","orange","yellow","green"])
plt.title("Distribution of Countries by Happiness Level")
plt.show()
```

# Obseravtion

From the pie chart, we can see that a large number of countries have average happiness levels,
while fewer countries fall into the good, low, or poor categories. This suggests that most countries have moderate happiness scores.
